### Overview & Objectives
The goal of this project is to build an end-to-end Machine Learning pipeline to classify iris flowers into three distinct species (**Iris-setosa**, **Iris-versicolor**, and **Iris-virginica**) based on four morphological measurements:
1. `SepalLengthCm`: Sepal length in centimeters
2. `SepalWidthCm`: Sepal width in centimeters
3. `PetalLengthCm`: Petal length in centimeters
4. `PetalWidthCm`: Petal width in centimeters

In this initial exploration notebook (**Monday deliverable**), we perform:
- Loading the Kaggle UCI Iris dataset from `data/iris-dataset.csv`
- Dropping the extraneous `Id` row identifier column
- Analyzing structural properties: `.shape`, `.dtypes`, `.info()`
- Computing statistical summaries: `.describe()`
- Verifying missing values and inspecting class distribution and balance


## 1. Environment Setup & Data Loading
We import core analytical libraries and resolve file paths dynamically to support execution whether run from the `week-7/notebooks/` directory or root.


```python
from pathlib import Path
import pandas as pd
import numpy as np

# Configure path dynamically to dataset
data_path = Path("../data/iris-dataset.csv")
if not data_path.exists():
    data_path = Path("week-7/data/iris-dataset.csv")

# Load raw dataset into a Pandas DataFrame
df = pd.read_csv(data_path)
print(f"Dataset successfully loaded from: {data_path}")
print(f"Raw shape: {df.shape}")
df.head()
```

## 2. Data Cleaning: Dropping the `Id` Column
The `Id` column is an arbitrary row index assigned in the Kaggle UCI dataset. It carries no biological signal and would introduce spurious correlation if retained. We drop it immediately.


```python
# Drop non-informative Id column
if "Id" in df.columns:
    df.drop(columns=["Id"], inplace=True)
    print("Column 'Id' dropped successfully.")

print(f"New DataFrame dimensions: {df.shape}")
df.head()
```

## 3. Structural Properties & Data Types
We inspect dimensions, data types, and non-null counts using `.shape`, `.dtypes`, and `.info()`.


```python
print("--- Dataset Shape ---")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}
")

print("--- Column Data Types ---")
print(df.dtypes)
print("
--- Concise DataFrame Summary ---")
df.info()
```

## 4. Descriptive Summary Statistics
We compute summary metrics (mean, standard deviation, percentiles, minimum, and maximum) for all four numerical features.


```python
# Summary statistics for numerical features
df.describe()
```

### Per-Species Mean & Standard Deviation
Breaking down feature means by species provides initial quantitative evidence of species divergence.


```python
# Per-species feature means
df.groupby("Species").mean()
```

## 5. Missing Value & Data Integrity Verification
We check for null, missing, or NaN values across every feature.


```python
# Check for missing values
missing_counts = df.isnull().sum()
print("Missing values per column:")
print(missing_counts)
print(f"Total missing values in entire dataset: {missing_counts.sum()}")
```

## 6. Class Distribution & Balance Verification
We verify the sample count and frequency distribution for each of the three target species.


```python
# Value counts and percentages per species
species_counts = df["Species"].value_counts()
species_pct = df["Species"].value_counts(normalize=True) * 100

distribution_df = pd.DataFrame({
    "Sample Count": species_counts,
    "Percentage (%)": species_pct
})
print("Class Distribution:")
distribution_df
```
